```python
import pandas as pd
import matplotlib.pyplot as plt
 
df = pd.read_csv("students_clean_p1.csv")

plt.style.use("seaborn-v0_8-whitegrid")

course_mean = df.groupby("course")["score"].mean().sort_index()
plt.plot(course_mean.index, course_mean.values, marker="o", label="Средний балл")
plt.xlabel("Курс"); plt.ylabel("Средний балл")
plt.title("Средний балл по курсам (Mathplotlib)"); plt.legend(); plt.show()
 
city_counts = df["city"].value_counts()
plt.bar(city_counts.index, city_counts.values)
plt.title("Количество студентов по городам (Mathplotlib)"); plt.show()
 
plt.hist(df["score"], bins=8)
plt.hist(df["score"], bins=8, color="#4EC456E3", edgecolor="white")
plt.title("Распределение оценок (Mathplotlib)"); plt.show()
 
plt.scatter(df["study_hours"], df["score"])
plt.title("Подготовка и оценка (Mathplotlib)"); plt.show()
 
plt.boxplot(df["score"])
plt.title("Boxplot оценок (Mathplotlib)"); plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x="score", kde=True)
plt.title("Гистограмма с KDE (Seaborn)")
plt.xlabel("Оценка"); plt.ylabel("Количество студентов")
plt.show()

sns.boxplot(data=df, x="city", y="score")
plt.title("Boxplot оценок по городам (Seaborn)")
plt.xlabel("Город"); plt.ylabel("Оценка")
plt.xticks(rotation=30)
plt.show()

sns.scatterplot(data=df, x="study_hours", y="score", hue="city")
plt.title("Scatter с hue по городам (Seaborn)")
plt.xlabel("Часы подготовки"); plt.ylabel("Оценка")
plt.show()

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Тепловая карта корреляций (Seaborn)")
plt.show()

pp = sns.pairplot(df[["age","study_hours","attendance","score"]])
pp.fig.suptitle("Pairplot числовых признаков (Seaborn)", y=1.02)
plt.show()
```

```python
import plotly.express as px

course_mean = df.groupby("course")["score"].mean().sort_index()
city_counts = df["city"].value_counts()

fig = px.line(x=course_mean.index, y=course_mean.values, markers=True,
              title="Средний балл по курсам (Plotly)")
fig.show()

fig = px.bar(x=city_counts.index, y=city_counts.values,
             title="Количество студентов по городам (Plotly)")
fig.show()

fig = px.scatter(df, x="study_hours", y="score", color="city",
                  hover_name="name", title="Интерактивный анализ успеваемости (Plotly)")
fig.show()
```
