```python
import pandas as pd
import numpy as np
 
df = pd.read_csv("students_data.csv")
 
print(df.head())
print(df.tail())
print(df.shape)
print(df.dtypes)
df.info()
print(df.describe())
```

```python
import pandas as pd
import numpy as np
 
print(df.isnull().sum())
 
df["age"] = df["age"].fillna(df["age"].median())
df["study_hours"] = df["study_hours"].fillna(df["study_hours"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].mean())
df["score"] = df["score"].fillna(df["score"].mean())

print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np
 

 
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df.shape)

```

```python
df.to_csv("students_clean_p1.csv", index=False)
```

```python
import pandas as pd
import numpy as np
 

high_score = df[df["score"] > 80]
 
city_score = df.groupby("city")["score"].mean().sort_values(ascending=False)
course_att = df.groupby("course")["attendance"].mean().sort_values(ascending=False)
 
df_sorted = df.sort_values("score", ascending=False)
 
df["high_score"] = df["score"] >= 80

print(high_score)
print(city_score)
print(course_att)
print(df_sorted.head())
```
