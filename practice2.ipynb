```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
 
df = pd.read_csv("students_clean_p1.csv")
 
print(df.shape)
print(df.describe())
print(df.isnull().sum())
print(df["city"].value_counts())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

sns.histplot(data=df, x="score", kde=True)
plt.title("Распределение итоговых оценок")
plt.xlabel("Оценка"); plt.ylabel("Количество студентов")
plt.show()
 
sns.boxplot(data=df, y="score")
plt.title("Boxplot оценок студентов")
plt.show()

sns.boxplot(data=df, y="attendance")
plt.title("Boxplot посещаемости")
plt.show()
 
sns.scatterplot(data=df, x="study_hours", y="score", hue="gender")
plt.title("Связь часов подготовки и оценки")
plt.show()
 
df["city"].value_counts().plot(kind="bar")
plt.title("Распределение студентов по городам")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Корреляционная матрица")
plt.show()
```
